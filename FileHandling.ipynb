### Opening a file
Files in python can be opened using `open()` function.

```python
File = open ( "Sample.txt" )
```

Now, that we've opened the file successfully let's verify it

```python
type (File)
```

In python if we open the file it's crucial that you close it too once operations on file are performed to do that we use close fucntion, but for now we will verify first if the file is closed or not?

```python
File.closed
```

If file was closed it could've returned false.

```python
# To read the file content
File.read()
```

As discussed once we're done with the file we have to close it, 

Question should be why? 
If you know the answer huge applause for you if not then let's suppose you've opened a huge log file, if not closed after operations you will encouter your program eating lot of memory for no reason plus file will be prone to the modification if opened in write/append mode.

So to do that...

```python
print(File.closed)

if File.closed == False:
    # Closes the file
    File.close()
    print ('File Closed')

# Verify
print(File.closed)
```

## File  Modes

- 'r' : open for reading (default)
- 'w' : open for writing, truncating the file first
- 'x' : create a new file and open it for writing
- 'a' : open for writing, appending to the end of the file if it exists
- 'b' : binary mode
- 't' : text mode (default)
- '+' : open a disk file for updating (reading and writing)

--- 

But Commonly we use below modes

---

- `r`  : Read Only
- `w`  : Write Only
- `a`  : Append data at EOF (end of file)
- `wt` : Write text
- `wb` : Write Binary
- `rb` : Read Binary
- `rt` : Read Text

Notice no file can be opened in both the mode

## Aother method to open files using `with`

```python
with open('Sample.txt') as File:
    print(File.read())
```

## Writing on a file

Python supports writing in a file using `w` mode. this mode writes on a file in two ways...

- If file is present, insert the new data by **DELETING** the previous data
- If file is not present, Create it, insert the data from begining

```python
with open ('newFile.txt', 'w') as File:
    File.write('This is a sample text')
```

Now, let's see both behaviour

```python
with open ('newFile.txt', 'w') as File:
    File.write('This is a NEXT text after file was created')
```

## Handling the binary data
Python only allows two modes to handle the binary data. E.g. read mode and write mode

So let's say you've a image and want to write that imgae to another file, how could you do that?

```python
with open('python.png', 'rb') as file:
    image = file.read()

    with open('new_image.png', 'wb') as new_image:
        new_image.write(image)
```

Similarly, instead of using 'w' we can use 'a' mode to append the data to EOF of a given file.

```python
with open('Sample.txt', 'a') as File:
    File.write('\nThis line is appended afterwards\n')

with open('Sample.txt') as File:
    print(File.read())
```
