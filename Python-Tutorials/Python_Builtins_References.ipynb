# Master Guide to Python Built-in Functions

## 1. Iteration, Sequences & Collections
Tools for looping, processing sequences, and managing core iterables.

### `enumerate()`
* **Syntax:** `enumerate(iterable, start=0)`
* **Arguments:** 
  * `iterable`: Any sequence or object that supports iteration.
  * `start` (Optional): The starting index base. Defaults to `0`.
* **Description:** Adds a counter to an iterable and returns it as an enumerate object containing pairs of indices and items.

```python
fruits = ['apple', 'banana', 'cherry']
for index, fruit in enumerate(fruits, start=1):
    print(f"{index}: {fruit}")
```

### `zip()`
* **Syntax:** `zip(*iterables, strict=False)`
* **Arguments:**
  * `*iterables`: Multiple iterables to be aggregated together.
  * `strict` (Optional): If `True`, raises an error if the iterables are unequal in length.
* **Description:** Pairs elements from corresponding positions across multiple sequences into tuples.

```python
names = ['Alice', 'Bob']
scores = [85, 92]
print(list(zip(names, scores)))
```

### `filter()`
* **Syntax:** `filter(function, iterable)`
* **Arguments:**
  * `function`: A function that evaluates each item to a boolean. If `None`, extracts items that are truthy.
  * `iterable`: The collection being filtered.
* **Description:** Filters elements from an iterable based on whether they pass a conditional test function.

```python
numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)
```

### `map()`
* **Syntax:** `map(function, *iterables)`
* **Arguments:**
  * `function`: The transformation function to apply to each element.
  * `*iterables`: One or more sequences matching the function's arguments.
* **Description:** Applies a given function to all items in an input iterable.

```python
words = ['hello', 'world']
uppercase_words = list(map(str.upper, words))
print(uppercase_words)
```

### `range()`
* **Syntax:** `range(start, stop[, step])` or `range(stop)`
* **Arguments:**
  * `start` (Optional): Starting integer (inclusive).
  * `stop`: Upper ceiling bound integer (exclusive).
  * `step` (Optional): Increment step size value between calculations.
* **Description:** Generates an immutable numeric sequence commonly used in loops.

```python
print(list(range(0, 10, 2)))
```

### `sorted()`
* **Syntax:** `sorted(iterable, key=None, reverse=False)`
* **Arguments:**
  * `iterable`: The dataset to be organized.
  * `key` (Optional): A function acting as a custom sorting criteria.
  * `reverse` (Optional): If set to `True`, sorts descending.
* **Description:** Returns a completely new sorted list from the items in any iterable.

```python
unordered = [5, 1, 4, 2]
print(sorted(unordered, reverse=True))
```

### `reversed()`
* **Syntax:** `reversed(seq)`
* **Arguments:**
  * `seq`: An object that supports the sequence protocol or has a `__reversed__()` method.
* **Description:** Returns a reverse iterator over a sequence.

```python
items = ['a', 'b', 'c']
print(list(reversed(items)))
```

### `len()`
* **Syntax:** `len(s)`
* **Arguments:**
  * `s`: A sequence (string, bytes, tuple, list, range) or collection (dict, set, frozenset).
* **Description:** Returns the total number of items (length) in an object.

```python
print(len([10, 20, 30, 40]))
```

### `iter()` & `next()`
* **Syntax:** `iter(object)` / `next(iterator[, default])`
* **Arguments:**
  * `object`: An iterable collection support target.
  * `iterator`: An active iterable pointer reference state variable.
  * `default` (Optional): Returned value if the iterator is exhausted instead of raising `StopIteration`.
* **Description:** `iter()` extracts an internal iterator tracking sequence object. `next()` retrieves successive individual steps from it manually.

```python
my_list = ['x', 'y']
iterator = iter(my_list)
print(next(iterator))
print(next(iterator))
print(next(iterator, 'End reached'))
```

## 2. Math, Numeric & Reductions
Mathematical calculations, numerical property checks, and group analysis tools.

### `abs()`
* **Syntax:** `abs(x)`
* **Arguments:**
  * `x`: An integer, floating-point number, or an object implementing `__abs__()`.
* **Description:** Returns the absolute positive magnitude value of a number.

```python
print(abs(-5.6))
```

### `sum()`
* **Syntax:** `sum(iterable, start=0)`
* **Arguments:**
  * `iterable`: Collection containing numerical entries.
  * `start` (Optional): A value added to the final sum of elements. Defaults to `0`.
* **Description:** Sums up the total items found across an iterable object sequence.

```python
print(sum([1, 2, 3], start=10))
```

### `max()` & `min()`
* **Syntax:** `max(iterable, *[, key, default])` / `min(iterable, *[, key, default])`
* **Arguments:**
  * `iterable`: The target tracking evaluation sequence.
  * `key` (Optional): One-argument filtering criteria matching reference parameters.
  * `default` (Optional): Default return value if the provided iterable is completely empty.
* **Description:** Locates the highest and lowest items across values or structures respectively.

```python
vals = [10, 4, 99, 2]
print("Max:", max(vals), "Min:", min(vals))
```

### `divmod()`
* **Syntax:** `divmod(a, b)`
* **Arguments:**
  * `a`: The dividend numerator value.
  * `b`: The divisor denominator value.
* **Description:** Returns a tuple containing both the floor quotient and the remainder calculation simultaneously.

```python
quotient, remainder = divmod(23, 5)
print(f"Quotient: {quotient}, Remainder: {remainder}")
```

### `pow()`
* **Syntax:** `pow(base, exp, mod=None)`
* **Arguments:**
  * `base`: Base primary evaluation element.
  * `exp`: Power exponent indicator element.
  * `mod` (Optional): Computes result modulo `mod` efficiently under the hood if present.
* **Description:** Returns base raised to the power exponent ($base^{exp}$).

```python
print(pow(2, 3))       # 2**3
print(pow(2, 3, 5))    # (2**3) % 5
```

### `round()`
* **Syntax:** `round(number, ndigits=None)`
* **Arguments:**
  * `number`: Floating-point entry.
  * `ndigits` (Optional): Decimal point truncation precision limit integer.
* **Description:** Rounds a floating-point number to specified decimal precision places.

```python
print(round(3.14159, 3))
```

### `bin()`, `hex()`, `oct()`
* **Syntax:** `bin(x)` / `hex(x)` / `oct(x)`
* **Arguments:**
  * `x`: Integer value to convert.
* **Description:** Transforms integer inputs to their equivalent binary (`0b`), hexadecimal (`0x`), or octal (`0o`) string representations.

```python
num = 42
print(bin(num), hex(num), oct(num))
```

### `hash()`
* **Syntax:** `hash(object)`
* **Arguments:**
  * `object`: Immutable types supporting the hashing protocol.
* **Description:** Generates an integer hash value utilized for speedy dictionary lookup tasks internally.

```python
print(hash("sample_string"))
```

## 3. Type Conversion & Object Instantiation
Constructors for type-casting or explicitly generating Python structures.

### Core Primitive Typings
* **Methods:** `int()`, `float()`, `complex()`, `bool()`, `str()`
* **Syntax Example:** `int(value, base=10)` / `str(object='')`
* **Description:** Standard scalar type constructors converting data between numbers, text, or truthy states.

```python
print(int("101", base=2))
print(float("4.5"))
print(bool(0), bool(99))
print(str(123))
```

### Collection Initializers
* **Methods:** `list()`, `tuple()`, `dict()`, `set()`, `frozenset()`, `object()`
* **Description:** Creates data container types. `frozenset()` creates an unmodifiable immutable set, while `object()` returns a completely blank base structure component instance.

```python
print(list((1, 2, 3)))
print(set([1, 1, 2]))
print(frozenset([1, 2, 3]))
```

### Binary Data Handlers
* **Methods:** `bytes()`, `bytearray()`, `memoryview()`
* **Description:** Manipulates raw binary data buffers. `bytes()` is immutable; `bytearray()` supports in-place structural item edits; `memoryview()` allows working directly with an object's memory buffers without making a costly copy.

```python
b_arr = bytearray([65, 66, 67])
b_arr[0] = 68  # Mutable modification!
print(b_arr)
view = memoryview(bytes([1, 2, 3]))
print(view[0])
```

### String & Character Encoders
* **Methods:** `chr()`, `ord()`, `ascii()`, `repr()`
* **Description:** Code unit and visualization interpreters. `chr()` handles changing integer values back into matching code character strings, `ord()` converts individual characters to integers, `repr()` gives a string representation meant for developers, and `ascii()` escapes non-ASCII characters.

```python
print(chr(97))         # Unicode point to string character
print(ord('a'))        # String character to integer value
print(repr("line\n")) # Output preserving original literal representations
```

## 4. Logic, Status & Introspection
Conditional group validation patterns and structural type verification utilities.

### `all()` & `any()`
* **Syntax:** `all(iterable)` / `any(iterable)`
* **Description:** `all()` returns `True` if **every single item** inside evaluates to true (like a giant `and` chain). `any()` returns `True` if **at least one item** inside evaluates to true (like a giant `or` chain).

```python
flags = [True, False, True]
print("All items True?:", all(flags))
print("Any item True?:", any(flags))
```

### `isinstance()` & `issubclass()`
* **Syntax:** `isinstance(object, classinfo)` / `issubclass(class, classinfo)`
* **Description:** Validates types. `isinstance()` checks if an object is an instance of a class, whereas `issubclass()` checks class inheritance structures directly.

```python
print(isinstance("hello", str))
print(issubclass(bool, int)) # bool inherits properties from int class natively
```

### `callable()`
* **Syntax:** `callable(object)`
* **Description:** Checks if an object can be called like a function (returns `True` for functions, classes, methods, and instances with a `__call__()` method).

```python
print(callable(print))
print(callable(42))
```

### `type()`
* **Syntax:** `type(object)`
* **Description:** Returns the exact underlying class type of an object.

```python
print(type([]))
```

## 5. Reflection, Namespace & Dynamic Execution
Advanced metadata inspectors, namespace dictionaries, and meta-programming modules.

### `dir()`, `id()` & `vars()`
* **Description:** Diagnostic helpers. `dir()` prints out all methods and attributes available on an object; `id()` yields its unique system memory address; `vars()` returns the underlying `__dict__` attribute storage dictionary mapping of a class/object.

```python
text = "test"
print("Memory ID:", id(text))
print("Attributes list snippet:", dir(text)[:3])
```

### `locals()` & `globals()`
* **Description:** Accesses active running environments mapping schemas directly. `locals()` collects variables defined in the current local block scope; `globals()` tracks variables in the current global module scope.

```python
temp_var = "Active inside scope state tracking"
print("Is variable in globals dictionary?:", 'temp_var' in globals())
```

### `eval()`, `exec()` & `compile()`
* **Description:** Evaluates strings as real Python code. `eval()` executes singular expression statements and returns a value; `exec()` runs complex multiline code blocks; `compile()` optimizes a code string into byte-code before execution.

```python
print(eval("3 + 5"))
exec("for i in range(2): print('Looping dynamically!')")
```

### Attribute Manipulation (`hasattr`, `getattr`, `setattr`, `delattr`)
* **Description:** Programmatically reads, writes, edits, or deletes an object's properties using strings.

```python
class Dummy: pass
d = Dummy()
setattr(d, 'status', 'Working fine!')
if hasattr(d, 'status'):
    print(getattr(d, 'status'))
    delattr(d, 'status')
```

### Object Oriented Decorators & Meta Framework tools
* **Methods:** `property()`, `classmethod()`, `staticmethod()`, `super()`, `__import__()`
* **Description:** Built-in tools for OOP architecture design. `super()` hooks up parent class methods, `@staticmethod` and `@classmethod` adjust method scoping bindings, and `__import__()` handles programmatic library imports.

```python
class MathTools:
    @staticmethod
    def add(x, y): return x + y
print(MathTools.add(5, 10))
```

Data input/output streams, string formatting, and local filesystem resource controls.

### `print()`
* **Syntax:** `print(*objects, sep=' ', end='\n', file=None, flush=False)`
* **Description:** Prints information to the standard output stream (usually the console).

```python
print("Item1", "Item2", sep=" | ", end=" !!!\n")
```

### `format()`
* **Syntax:** `format(value[, format_spec])`
* **Description:** Formats a value according to a specific format descriptor string (e.g., currency comma insertions or adding percentage conversions).

```python
print(format(1234567, ","))
print(format(0.725, "%"))
```

### `open()`

* **Description:** Opens a file and returns a corresponding file stream object for reading or writing.

```python
with open("sample_test_file.txt", "w") as f:
    f.write("Notebook file creation operational tracking validation line.")
with open("sample_test_file.txt", "r") as f:
    print(f.read())
```
