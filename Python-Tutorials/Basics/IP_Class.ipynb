```python
def Binary(num):
    
    binary_lst = list()
    
    while num != 0:
        binary = num % 2
        binary_lst.append(binary)
        num = num // 2
    
    binary_lst.reverse()

    binary_str = ''.join(map(str, binary_lst))

    return binary_str.zfill(8)

def IPv4ToBinary(IP):

    ip = IP.split('.')
    
    ip_binary = list()
       
    if len(ip) != 4 or not all ( 0 <= int(block) <= 255 for block in ip ) : 
        print ("Error!!! invalid IP")
        
    else:
        
        for iteration in ip:
            binary = Binary(int(iteration))
            ip_binary.append(binary)
        
        print ('.'.join(ip_binary))

# IPv4ToBinary("192.168.0.45")
```
