## Constructors

Constructor is a special method that runs automatically by calling method `__init__` when an object is created, Basically it's responsible for creating an object and allocating the data /attributes to (initializing) it. 

Basically, 
1. It creates memory for the object
2. Initializes it's attributes

Without Constructors, If we try to create an object, it will give error as below:

````python
class Car:
    pass

bmw = Car()

print(bmw.brand)
````

Now, that we've an object's blueprint, We also need a data (or attributes) to define it's various properties such as the color, variant, moodel, type, etc. 

````python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
````

### The Initialization of object 
Notice in above snippet we've *`__init__`* an special python method with soul purpose to initialize an object after it is created by assigining initial values to it. 

Say in our example

````python
bmw = Car("BMW", "M5 Competition", 2025)
````

Python automatically runs the `__init__(...)` to store 

```python
brand = "BMW" 
model = "M5 Competition"
year  = 2025
```

#### The *`self`* game

Notice in the same snippet, there's an additional keyword argument *`self`* was passed along with the 'brand', 'model', 'year'. Here, self refers to the current object.

Suppose,

````python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
````

Python secretly does... 
```python 
Car.__init__(bmw, "BMW")
```

So, `self` becomes 'bmw'

Translating 

```python
self.brand  = brand
```
into,
```python
bmw.brand = "BMW"
```


Here's How it works, Suppose for below code snippet...


##  Destructor

#### What is a destructor?
    
A destructor is a special method called using `__del__()` that Python calls when an object is about to be destroyed. It is mainly used for cleanup tasks such as releasing resources, closing files, or closing database connections. However, its execution time is not guaranteed.

Unlike contructors destructors are not used very often as python alraedy has it's own garbacge collection mechanish that decides when to remove and object from memory. And by it's nature, as long as even a single reference is present for a particular object it python does not delete the object from a memory it just deletes the particular reference for a particular object. 

Example.

````python
class Car:
    def __init__(self):
        self.friend = None
    
    def __del__(self):
        print("Destroyed")

car1 = Car()

car2 = Car()

car1.friend = car2

car2.friend = car1
````

````python
del car1
del car2
````

As long as even a single reference is present for a object (in above case it's a circular reference) It does may delay the cleanup. Determining the time it wil take to clean up is undefined as program may can crash before it happens. 

Now the question rises if we want to clean up the resourses how can we do it? 

So the answer is simple, Python also provides the way to explicity deal with such situations...

````python
class Car:
    def close(self):
        close_operation() # It can be renamed 
````

Basically, `__del__()` is not guaranteed to execute at a specific time because Python uses garbage collection. An object is destroyed only when Python determines that it is no longer needed, and the exact timing depends on references, circular references, interpreter shutdown, and the garbage collector's behavior. Therefore, important cleanup logic should not rely on `__del__()`.
