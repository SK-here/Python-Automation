## Constructors

Constructor is a special method that runs automatically by calling method `__init__` when an object is created, Basically it's responsible for creating an object and allocating the data /attributes to (initializing) it. 

Basically, 
1. It creates memory for the object
2. Initializes it's attributes

Without Constructors, If we try to create an object, it will give error as below:

````python
class Car:
    pass

bmw = Car()

print(bmw.brand)
````

Now, that we've an object's blueprint, We also need a data (or attributes) to define it's various properties such as the color, variant, moodel, type, etc. 

````python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
````

##  Destructor

#### What is a destructor?
    
A destructor is a special method called using `__del__()` that Python calls when an object is about to be destroyed. It is mainly used for cleanup tasks such as releasing resources, closing files, or closing database connections. However, its execution time is not guaranteed.

Unlike contructors destructors are not used very often as python alraedy has it's own garbacge collection mechanish that decides when to remove and object from memory. And by it's nature, as long as even a single reference is present for a particular object it python does not delete the object from a memory it just deletes the particular reference for a particular object. 

Example.

````python
class Car:
    def __init__(self):
        self.friend = None
    
    def __del__(self):
        print("Destroyed")

car1 = Car()

car2 = Car()

car1.friend = car2

car2.friend = car1
````

````python
del car1
del car2
````

As long as even a single reference is present for a object (in above case it's a circular reference) It does may delay the cleanup. Determining the time it wil take to clean up is undefined as program may can crash before it happens. 

Now the question rises if we want to clean up the resourses how can we do it? 

So the answer is simple, Python also provides the way to explicity deal with such situations...

````python
class Car:
    def close(self):
        close_operation() # It can be renamed 
````

Basically, `__del__()` is not guaranteed to execute at a specific time because Python uses garbage collection. An object is destroyed only when Python determines that it is no longer needed, and the exact timing depends on references, circular references, interpreter shutdown, and the garbage collector's behavior. Therefore, important cleanup logic should not rely on `__del__()`.

# Encapsulation

Now, extending further the exaple of `Car()`, Imagine we're engineers at the "BMW" building software for a BMW M5 Competition. Since it's a mechanical car, so there are some parts of it we need to expose enough, but not everything (especially design and implementation details).

This technique of hiding the implementation details and exposing only what's necessary is called as **Encapsulation** 

In other words
> Bundling the data (attributes) and it's behavhiour (methods) together and controlling how data can be accessed or modified  

In our Car analogy, let's say:

> - You can start the car
> - You can accelarate or deaccelarate the car
> - You can aply breaks
> - You can open/replace/repair few components of car 
> - Or even you can modify few parts of too

But, here's the catch (normall):

> - You cannot modify the ECU
> - You cannot modify the Feul injectors
> - You cannot directly alter the ABS Sensors
> - You cannot directly perform engine tuning
> - You cannot change the behaviour of sensors

Fo sure you can open the enginer and perform some of the repairs too but it also need to be done in a certain defined way (Unless you're a expert car mechanic with years of experience with same class of cars even still you cannot change certain things a good mechanic will know this 😌)

The car hasa exposed only the safe interface to you. Similarly,

Let's take an example

````python
# Speed of BMW
class Car:
    def __init__(self):
        self.speed = 0
````

Assuming that the code is exposed, some random developer

````python
bmw = Car()
bmw.speed = -60  
````

Now, the BMW Speed = -60 mph

### How encapsulation works?

Encapsulation has 3 levels
- pubic 
- private
- protected

#### Public
By default every class in python is public class

````python
class Car:
    def __init__(self):
        self.brand = "BMW"

# Accessible anywhere
bmw = Car()
bmw.brand
````

### Protected

Protected methods/attributes are supposed to be used inside the classes not outside of it.

In python we use `_` (underscore) as a prefix in method's or attribute names.

````python
class Car:
    def __init__(self):
        self._fuel = 60
````

Accessible anywhere but it's a developers agreement that says 'I trust you but do not touch this directly' However, It's not a restriction

````python
bmw = Car()
bmw._fuel
````

````python
bmw._fuel = 50

bmw._fuel
````

Unlike C++ or Java where protected methods or Data/attributes are not accessible outside the class, In python It's just a notation to point the developers that the method or attribute should not be used outside the class. However, Python does allows it. We've already seen in above example.

### Private

In python private methods/attributes denoted using `__` double underscores and it is also accessible like protected, However, python uses a trick called `name mangling` to at least disable direct access to the code.

Let's see this in action.

````python
class Car:
    def __init__(self):
        self.__engine_temp = 90
````

````python
bmw = Car()

bmw.__engine_temp # Attribute Error
````

Now, let's see name mangling in action...

```python
# Python changes the code from 
self.__engine_temp 

# to the code
self._Car__engine_temp
```

Disabling direct access to the method/attrib. But it can be accessed using

````python
bmw._Car__engine_temp
````

Now, let's try to modify it a bit

````python
bmw._Car__engine_temp = 95
bmw._Car__engine_temp
````

As you can see we've already modified the private method too... However, as a developers good practice it is highly advised not to do that since these notations are used industry wide and by doing to even as a experiment while working pn a Enterprise results will be unpredictable or entire might break.
