## Docstrings

Docstrings are used to document what a code snippet does it's denoted by `"""`...`"""`

````python
# This Function Greets
def greet():
    print("Hey, Have a good day!")
````

Notice above the comment does the same thing then why docstrings? Right... 

Let's see with example 
```python 
print()
```

````python
print?
````

***BUT***

````python
greet?
````

Notice the 'Docstring' field. 

Similarly for a large codebase, where you're writing  a reusable code, it's a best practive to give snippet a description to avoid any future confusions.

Let's see this in action

````python
# This Function Greets
def greet():
    """
    This function greets the user.
    """
    print("Hey, Have a good day!")
````

## Positioal arguments in Function

In function, the value passed when calling a function should match according to their positions otherwise results will be skewed 

Also, the number of arguments must be same as defined in functions.

Example:

````python
def subtract (num1, num2):
    return num1 - num2
````

Remember, results matter here...

````python
subtract(15, 12) # will be 3
````

````python
subtract(12,15) # will be -3
````

## Default Arguments

Arguments that can be given but are totally optional

````python
def Intro(Name, Hobby = None):
    return "Hey! My name is " + str(Name) + " and my hobby is " + str(Hobby) 
````

````python
Intro("SK")
````

**NOTE:** A non-default arguments follows Default arguments not vice-versa

Let's  have a look what it means

````python
# Notice the swapping or args here
# Will throw SyntaxError
def Intro(Hobby = None, Name):
    return "Hey! My name is " + str(Name) + " and my hobby is " + str(Hobby) 
````

````python
# Now again 
def Intro(Name, Hobby = None, Age = None):
    return "Hey! My name is " + str(Name) + ", my hobby is " + str(Hobby) + ", and my age is: " + str(Age) 
````

````python
Intro('SK')
````

***But...***

````python
# Notice the Age arg, it's not default -->>> Will Throw Error
def Intro(Name, Hobby = None, Age):
    return "Hey! My name is " + str(Name) + ", my hobby is " + str(Hobby) + ", and my age is: " + str(Age) 
````

## Arbitary Arguments

- A number of values you want to pass is unknown or maybe in some cases it's 'NONE' ex. print()

````python
def ArgFunc(*args):
    print(args)
````

````python
ArgFunc(2, 3, 4, 3-4, 4 - 3,"SK") 
````

````python
# Again,

def AnotherArgFunc(*args):
    print (type(args))
    for iteration in args:
        print(iteration, end=" ")
````

**NOTE:** The return type of `args` is `tuple` and the order of arguments matter here, Since there's no way to know which arguments is placed where...

````python
AnotherArgFunc(2, 3, 4, 3-4, 4 - 3,"SK") 
````

## Keyword Arguments

Unlike Arbitary arguments the positional args uses `variable=value` which then translated as a dictionary so the order of arguments doesn't matters.

````python
def add(**kwargs):
    print(type(kwargs))
    return kwargs
````

````python
add(a=2,b=3)
````

Let's see the diffrence between all the args in action.

````python
def Arguments(arg1, arg2, arg3 = None, *args, **kwargs):
    return arg1, arg2, arg3, args, kwargs
````

````python
Arguments(1,2)
````

Notice that we've given two args remaining were optional now, let's see what will happen if any of the arguments is not provided, will it work? 

````python
Arguments(1)
````

See the error, the fuction throws error are two args are not optional they must be provided.
