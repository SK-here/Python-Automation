## How to handle them all? 

Answer is `try` & `except` block:
- `try`: This block handles the erroe in code
- `except`: This block provides an aleternative way on how you want to handle the error for example: Printing and custom message for error.

### 1. Basic Try-Except Block Example
Let's catch a common runtime error like division by zero using a simple `try-except` setup.

````python
try:
    # Attempting to divide by zero
    result = 10 / 0
except ZeroDivisionError as e:
    # Catching the specific error and displaying a friendly message
    print(f"An error occurred: {e}")
    print("You cannot divide a number by zero!")
````

### 2. Handling Multiple Specific Exceptions
You can chain multiple `except` blocks to handle different errors elegantly based on what went wrong.

````python
try:
    # Change the value to test different exceptions
    user_input = "abc"
    number = int(user_input)
    
    # Indexing error simulation
    my_list = [1, 2, 3]
    print(my_list[5])
    
except ValueError:
    print("Caught a ValueError! You cannot convert text into an integer.")
except IndexError:
    print("Caught an IndexError! You tried to access an item out of bounds.")
````

### 3. Using the `else` and `finally` Clauses
- `else`: This block runs **only if** the code within the `try` block executed completely without throwing any exceptions.
- `finally`: This block runs **no matter what**, regardless of whether an error occurred or was caught. It's highly useful for cleaning up resources, like closing files.

````python
try:
    print("Opening operation framework...")
    data = 100 / 2
except ZeroDivisionError:
    print("Division failed!")
else:
    print(f"Success! The result is: {data}")
finally:
    print("Cleaning up and finalizing process... Completed.")
````

### 4. Manually Raising Exceptions (`raise`)
Sometimes, you need to deliberately trigger an exception when your custom constraints or business logic parameters are violated.

````python
def check_user_age(age):
    if age < 0:
        raise ValueError("Age cannot be a negative value!")
    return f"Valid age: {age}"

try:
    print(check_user_age(-5))
except ValueError as custom_error:
    print(f"Validation Exception Triggered: {custom_error}")
````
